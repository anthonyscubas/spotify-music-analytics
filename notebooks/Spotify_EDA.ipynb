# Spotify | Análisis exploratorio reproducible

Este notebook se ejecuta con la muestra ficticia incluida. Para el conjunto original, configura `DATA_PATH = ROOT / "data" / "spotify_data.csv"` tras obtenerlo legalmente. No representa métricas en tiempo real de Spotify.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / "spotify_eda").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from spotify_eda.analysis import load_tracks, summary, top_artists, top_tracks, genres, correlations
from spotify_eda.reporting import create_report
DATA_PATH = ROOT / "data" / "spotify_demo.csv"
df = load_tracks(DATA_PATH)
summary(df)
```

## Calidad y límites
Se eliminan filas sin identificador, nombre de canción o artista; se deduplican IDs; las métricas con valores inválidos se marcan como ausentes. Los géneros pueden tener varias etiquetas por registro. Una mayor puntuación no prueba causalidad.

```python
df.isna().sum().sort_values(ascending=False).head(12)
```

## Rankings y correlaciones
Se exige un mínimo de cinco canciones por artista para comparar promedios, y se evita comparar promedios basados en una única canción.

```python
display(top_artists(df, min_tracks=5)); display(top_tracks(df)); display(genres(df)); display(correlations(df))
```
